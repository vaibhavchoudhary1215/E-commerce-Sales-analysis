```python
rfm_data = Ecan[Ecan['Order_Status'] == 'Delivered'].copy()
```

```python
# Convert date
Ecan['Order_Date'] = pd.to_datetime(
    Ecan['Order_Date'],
    dayfirst=True
)

# Use delivered orders for RFM
rfm_data = Ecan[
    Ecan['Order_Status'] == 'Delivered'
].copy()

# Reference date = one day after the latest order
reference_date = rfm_data['Order_Date'].max() + pd.Timedelta(days=1)

# Calculate RFM
rfm = rfm_data.groupby('Customer_ID').agg(
    Recency=('Order_Date',
             lambda x: (reference_date - x.max()).days),

    Frequency=('Order_ID', 'nunique'),

    Monetary=('Sales', 'sum')
).reset_index()

print(rfm.head())
```

```python
# Score ---
rfm['R_Score'] = pd.qcut(
    rfm['Recency'],
    5,
    labels=[5, 4, 3, 2, 1]
)
```

```python
# Frequency ---
rfm['F_Score'] = pd.qcut(
    rfm['Frequency'].rank(method='first'),
    5,
    labels=[1, 2, 3, 4, 5]
)
```

```python
# Monetary --- 
rfm['M_Score'] = pd.qcut(
    rfm['Monetary'],
    5,
    labels=[1, 2, 3, 4, 5]
)
```

```python
# Converting in integer ---
rfm[['R_Score', 'F_Score', 'M_Score']] = (
    rfm[['R_Score', 'F_Score', 'M_Score']]
    .astype(int)
)
```

```python
# RFM score
rfm['RFM_Score'] = (
    rfm['R_Score'].astype(str) +
    rfm['F_Score'].astype(str) +
    rfm['M_Score'].astype(str)
)

print(rfm.head())
```

```python
# Create customer segments
def segment_customer(row):

    R = row['R_Score']
    F = row['F_Score']
    M = row['M_Score']

    if R >= 4 and F >= 4 and M >= 4:
        return 'Champions'

    elif R >= 3 and F >= 3:
        return 'Loyal'

    elif R >= 4 and F <= 2:
        return 'New Customers'

    elif R <= 2 and F >= 3:
        return 'At Risk'

    else:
        return 'Lost'


rfm['Segment'] = rfm.apply(
    segment_customer,
    axis=1
)
```

```python
#  Plotting ---
plt.figure(figsize=(12, 7))

sns.scatterplot(
    data=rfm,
    x='Recency',
    y='Frequency',
    hue='Segment',
    size='Monetary',
    sizes=(50, 600),
    alpha=0.75
)

plt.title('RFM Customer Segmentation')
plt.xlabel('Recency (days since last purchase)')
plt.ylabel('Frequency (number of orders)')

plt.legend(
    bbox_to_anchor=(1.02, 1),
    loc='upper left'
)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# E-COMMERCE SALES & PROFITABILITY DASHBOARD
# ============================================================

# 1. DATA PREPARATION


Ecan['Order_Date'] = pd.to_datetime(
    Ecan['Order_Date'],
    dayfirst=True
)

# Use Delivered orders for actual completed sales
data = Ecan[
    Ecan['Order_Status'] == 'Delivered'
].copy()


# 2. KPI CALCULATIONS

total_sales = data['Sales'].sum()
total_profit = data['Profit'].sum()
total_orders = data['Order_ID'].nunique()
total_customers = data['Customer_ID'].nunique()

profit_margin = (total_profit / total_sales) * 100



# 3. MONTHLY SALES

monthly_sales = (
    data.groupby(
        data['Order_Date'].dt.to_period('M')
    )['Sales']
    .sum()
    .reset_index()
)

monthly_sales['Order_Date'] = (
    monthly_sales['Order_Date'].dt.to_timestamp()
)



# 4. CATEGORY SALES


category_sales = (
    data.groupby('Product_Category')['Sales']
    .sum()
    .sort_values(ascending=False)
)



# 5. REGIONAL PROFIT


region_profit = (
    data.groupby('Region')['Profit']
    .sum()
    .sort_values(ascending=False)
)



# 6. PAYMENT METHOD


payment_count = (
    data['Payment_Method']
    .value_counts()
)



# 7. RFM ANALYSIS


reference_date = (
    data['Order_Date'].max()
    + pd.Timedelta(days=1)
)

rfm = data.groupby('Customer_ID').agg(
    
    Recency=(
        'Order_Date',
        lambda x: (
            reference_date - x.max()
        ).days
    ),
    
    Frequency=(
        'Order_ID',
        'nunique'
    ),
    
    Monetary=(
        'Sales',
        'sum'
    )

).reset_index()



# RFM SCORES


rfm['R_Score'] = pd.qcut(
    rfm['Recency'],
    5,
    labels=[5, 4, 3, 2, 1],
    duplicates='drop'
)

rfm['F_Score'] = pd.qcut(
    rfm['Frequency'].rank(method='first'),
    5,
    labels=[1, 2, 3, 4, 5],
    duplicates='drop'
)

rfm['M_Score'] = pd.qcut(
    rfm['Monetary'],
    5,
    labels=[1, 2, 3, 4, 5],
    duplicates='drop'
)

rfm[
    ['R_Score', 'F_Score', 'M_Score']
] = rfm[
    ['R_Score', 'F_Score', 'M_Score']
].astype(int)



# CUSTOMER SEGMENTS


def segment_customer(row):

    R = row['R_Score']
    F = row['F_Score']
    M = row['M_Score']

    if R >= 4 and F >= 4 and M >= 4:
        return 'Champions'

    elif R >= 3 and F >= 3:
        return 'Loyal'

    elif R >= 4 and F <= 2:
        return 'New Customers'

    elif R <= 2 and F >= 3:
        return 'At Risk'

    else:
        return 'Lost'


rfm['Segment'] = rfm.apply(
    segment_customer,
    axis=1
)


# 8. CREATE DASHBOARD


fig = plt.figure(
    figsize=(18, 12)
)

fig.suptitle(
    'E-Commerce Sales & Profitability Dashboard',
    fontsize=22,
    fontweight='bold'
)

gs = fig.add_gridspec(
    4,
    4,
    height_ratios=[0.8, 2, 2, 3]
)



# KPI CARDS


kpis = [
    ('Total Sales', f'₹{total_sales:,.0f}'),
    ('Total Profit', f'₹{total_profit:,.0f}'),
    ('Total Orders', f'{total_orders:,}'),
    ('Customers', f'{total_customers:,}')
]

for i, (title, value) in enumerate(kpis):

    ax = fig.add_subplot(
        gs[0, i]
    )

    ax.axis('off')

    ax.text(
        0.5,
        0.65,
        title,
        ha='center',
        fontsize=12
    )

    ax.text(
        0.5,
        0.25,
        value,
        ha='center',
        fontsize=18,
        fontweight='bold'
    )



# MONTHLY SALES TREND


ax1 = fig.add_subplot(
    gs[1, :2]
)

sns.lineplot(
    data=monthly_sales,
    x='Order_Date',
    y='Sales',
    marker='o',
    ax=ax1
)

ax1.set_title(
    'Monthly Sales Trend',
    fontweight='bold'
)

ax1.set_xlabel('Month')
ax1.set_ylabel('Sales')

ax1.tick_params(
    axis='x',
    rotation=45
)

ax1.grid(True)



# CATEGORY SALES


ax2 = fig.add_subplot(
    gs[1, 2:]
)

category_sales.sort_values().plot(
    kind='barh',
    ax=ax2
)

ax2.set_title(
    'Sales by Category',
    fontweight='bold'
)

ax2.set_xlabel('Sales')
ax2.set_ylabel('Category')



# REGIONAL PROFIT


ax3 = fig.add_subplot(
    gs[2, :2]
)

region_profit.plot(
    kind='bar',
    ax=ax3
)

ax3.set_title(
    'Profit by Region',
    fontweight='bold'
)

ax3.set_xlabel('Region')
ax3.set_ylabel('Profit')

ax3.tick_params(
    axis='x',
    rotation=0
)



# PAYMENT METHOD


ax4 = fig.add_subplot(
    gs[2, 2:]
)

payment_count.plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90,
    ax=ax4
)

ax4.set_title(
    'Orders by Payment Method',
    fontweight='bold'
)

ax4.set_ylabel('')



# RFM CUSTOMER SEGMENTATION


ax5 = fig.add_subplot(
    gs[3, :]
)

sns.scatterplot(
    data=rfm,
    x='Recency',
    y='Frequency',
    hue='Segment',
    size='Monetary',
    sizes=(40, 500),
    alpha=0.75,
    ax=ax5
)

ax5.set_title(
    'RFM Customer Segmentation',
    fontsize=14,
    fontweight='bold'
)

ax5.set_xlabel(
    'Recency (Days Since Last Purchase)'
)

ax5.set_ylabel(
    'Frequency (Number of Orders)'
)

ax5.legend(
    bbox_to_anchor=(1.02, 1),
    loc='upper left'
)



# FINAL FORMATTING


plt.tight_layout(
    rect=[0, 0, 1, 0.96]
)

plt.show()
```
